```python
import tensorflow as tf
from tensorflow import keras

# ------------------------------------------------------------------
# 1. Load data (same as your existing pipeline)
# ------------------------------------------------------------------
train_dir = "/Users/sanjib700/Desktop/My_Projects/Covide_image/Covid19-dataset/train"
test_dir  = "/Users/sanjib700/Desktop/My_Projects/Covide_image/Covid19-dataset/test"

IMG_SIZE = (150, 150)
BATCH_SIZE = 16

train_ds = keras.utils.image_dataset_from_directory(
    train_dir,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=True,
    seed=42,
)

test_ds = keras.utils.image_dataset_from_directory(
    test_dir,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False,
)

class_names = train_ds.class_names
num_classes = len(class_names)
print("Classes:", class_names)

# Rescale pixels 0-255 -> 0-1
normalization_layer = keras.layers.Rescaling(1.0 / 255)
train_ds = train_ds.map(lambda x, y: (normalization_layer(x), y))
test_ds = test_ds.map(lambda x, y: (normalization_layer(x), y))

# ------------------------------------------------------------------
# 2. Build a simple Sequential CNN
# ------------------------------------------------------------------
model = keras.Sequential([
    keras.layers.Input(shape=(150, 150, 3)),   # 150x150 image, 3 color channels

    # --- Convolutional block 1: learns simple patterns (edges, textures) ---
    keras.layers.Conv2D(16, (3, 3), activation="relu"),
    keras.layers.MaxPooling2D(2, 2),

    # --- Convolutional block 2: combines simple patterns into more complex shapes ---
    keras.layers.Conv2D(32, (3, 3), activation="relu"),
    keras.layers.MaxPooling2D(2, 2),

    # --- Convolutional block 3: even more complex shapes ---
    keras.layers.Conv2D(64, (3, 3), activation="relu"),
    keras.layers.MaxPooling2D(2, 2),

    # --- Flatten the 2D feature maps into a 1D list, like before ---
    keras.layers.Flatten(),

    # --- Dropout: randomly "switch off" 50% of neurons during training,
    #     to reduce overfitting -- important with only 251 training images ---
    keras.layers.Dropout(0.5),

    # --- Dense layers: same as your digits classifier ---
    keras.layers.Dense(64, activation="relu"),
    keras.layers.Dense(num_classes, activation="softmax"),  # 3 outputs: Covid, Normal, Viral Pneumonia
])

model.summary()

# ------------------------------------------------------------------
# 3. Compile
# ------------------------------------------------------------------
model.compile(
    loss="categorical_crossentropy",   # matches label_mode="categorical" (one-hot labels)
    optimizer="adam",
    metrics=["accuracy"],
)

# ------------------------------------------------------------------
# 4. Train
# ------------------------------------------------------------------
EPOCHS = 20
history = model.fit(
    train_ds,
    validation_data=test_ds,
    epochs=EPOCHS,
    verbose=2,
)

# ------------------------------------------------------------------
# 5. Evaluate
# ------------------------------------------------------------------
test_loss, test_acc = model.evaluate(test_ds, verbose=0)
print(f"\nTest accuracy: {test_acc:.4f}")

# ------------------------------------------------------------------
# 6. Try one prediction
# ------------------------------------------------------------------
import numpy as np
for images, labels in test_ds.take(1):
    preds = model.predict(images, verbose=0)
    for i in range(3):
        true_class = class_names[labels[i].numpy().argmax()]
        pred_class = class_names[preds[i].argmax()]
        confidence = preds[i].max()
        print(f"True: {true_class:16s} Predicted: {pred_class:16s} Confidence: {confidence:.1%}")
    break
```

```python
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix

# ------------------------------------------------------------------
# Run predictions on the ENTIRE test set (not just one batch)
# ------------------------------------------------------------------
all_true = []
all_pred = []

for images, labels in test_ds:
    preds = model.predict(images, verbose=0)
    all_true.extend(labels.numpy().argmax(axis=1))
    all_pred.extend(preds.argmax(axis=1))

all_true = np.array(all_true)
all_pred = np.array(all_pred)

print(f"Total test images checked: {len(all_true)}")
print()

# ------------------------------------------------------------------
# Per-class breakdown: precision, recall, f1-score
# ------------------------------------------------------------------
print("=== Per-class performance ===")
print(classification_report(all_true, all_pred, target_names=class_names))

# ------------------------------------------------------------------
# Confusion matrix: which classes get confused with which
# ------------------------------------------------------------------
print("=== Confusion matrix ===")
print("Rows = true class, Columns = predicted class")
cm = confusion_matrix(all_true, all_pred)
print("            ", "  ".join(f"{c[:8]:>8s}" for c in class_names))
for i, row in enumerate(cm):
    print(f"{class_names[i][:12]:12s}", "  ".join(f"{v:8d}" for v in row))
```

```python
import tensorflow as tf
from tensorflow import keras

# ------------------------------------------------------------------
# 1. Load data
# ------------------------------------------------------------------
train_dir = "/Users/sanjib700/Desktop/My_Projects/Covide_image/Covid19-dataset/train"
test_dir  = "/Users/sanjib700/Desktop/My_Projects/Covide_image/Covid19-dataset/test"

IMG_SIZE = (150, 150)
BATCH_SIZE = 16

train_ds = keras.utils.image_dataset_from_directory(
    train_dir,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=True,
    seed=42,
)

test_ds = keras.utils.image_dataset_from_directory(
    test_dir,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False,
)

class_names = train_ds.class_names
num_classes = len(class_names)
print("Classes:", class_names)

# Rescale pixels 0-255 -> 0-1
normalization_layer = keras.layers.Rescaling(1.0 / 255)
train_ds = train_ds.map(lambda x, y: (normalization_layer(x), y))
test_ds = test_ds.map(lambda x, y: (normalization_layer(x), y))

# ------------------------------------------------------------------
# 2. Build a simple Sequential model — Flatten + Dense layers only
#    (same style as the book's Fashion MNIST example, Chapter 10)
# ------------------------------------------------------------------
model = keras.Sequential([
    keras.layers.Input(shape=(150, 150, 3)),   # 150x150 image, 3 color channels
    keras.layers.Flatten(),                     # unroll into one long list of numbers
    keras.layers.Dense(300, activation="relu"), # hidden layer 1
    keras.layers.Dense(100, activation="relu"), # hidden layer 2
    keras.layers.Dense(num_classes, activation="softmax"),  # output: one neuron per class
])

model.summary()

# ------------------------------------------------------------------
# 3. Compile
# ------------------------------------------------------------------
model.compile(
    loss="categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"],
)

# ------------------------------------------------------------------
# 4. Train
# ------------------------------------------------------------------
EPOCHS = 20
history = model.fit(
    train_ds,
    validation_data=test_ds,
    epochs=EPOCHS,
    verbose=2,
)

# ------------------------------------------------------------------
# 5. Evaluate
# ------------------------------------------------------------------
test_loss, test_acc = model.evaluate(test_ds, verbose=0)
print(f"\nTest accuracy: {test_acc:.4f}")

# ------------------------------------------------------------------
# 6. Try a few predictions
# ------------------------------------------------------------------
for images, labels in test_ds.take(1):
    preds = model.predict(images, verbose=0)
    for i in range(3):
        true_class = class_names[labels[i].numpy().argmax()]
        pred_class = class_names[preds[i].argmax()]
        confidence = preds[i].max()
        print(f"True: {true_class:16s} Predicted: {pred_class:16s} Confidence: {confidence:.1%}")
    break
```
